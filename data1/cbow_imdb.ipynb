```python
import pandas as pd
import numpy as np
import re

from gensim.models import Word2Vec
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
df = pd.read_csv("/home/vansh-talyani/Downloads/LP IV Updated Datasets/CBOW/IMDB Dataset.csv")
```

```python
df.head()
```

```python
documents = df['review'].astype(str).tolist()
documents[0]
```

```python
labels = df['sentiment'].map({'positive':1,'negative':0}).values
labels
```

```python
sentences = [re.findall(r'\w+',text.lower()) for text in documents]
```

```python
sentences[0]
```

```python
x_train, x_test, y_train, y_test = train_test_split(sentences,labels,test_size=0.2,random_state=42)
```

```python
w2v = Word2Vec(x_train,vector_size=50,window=5,min_count=2,sg=0,epochs=20)
```

```python
def get_vectors(words):
    vectors = []
    for w in words:
        if w in w2v.wv:
            vectors.append(w2v.wv[w])

    if len(vectors)>0:
        return np.mean(vectors,axis=0)
    else:
        return np.zeros(50)

x_train = np.array([get_vector(s) for s in x_train])
x_test = np.array([get_vector(s) for s in x_test])
```

```python
model = Sequential()
model.add(Dense(128, activation='relu', input_shape=(50,)))
model.add(Dense(64, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
r= model.fit(x_train, y_train, epochs=20, batch_size=32)
```

```python
loss, accuracy = model.evaluate(x_test, y_test)

print("Loss:", loss)
print("Accuracy:", accuracy)
```
