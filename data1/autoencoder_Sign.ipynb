```python
import pandas as pd
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, UpSampling2D
from sklearn.metrics import confusion_matrix, classification_report
```

```python
train = pd.read_csv("/home/vansh-talyani/Downloads/LP IV Updated Datasets/AutoEncoder/Sign Language MNIST/sign_mnist_train.csv")
test = pd.read_csv("/home/vansh-talyani/Downloads/LP IV Updated Datasets/AutoEncoder/Sign Language MNIST/sign_mnist_test.csv")
```

```python
X_train = train.drop('label', axis=1).values
y_train = train['label'].values

X_test = test.drop('label', axis=1).values
y_test = test['label'].values

X_train
```

```python
X_train = X_train.reshape(-1, 28, 28, 1) / 255.0
X_test = X_test.reshape(-1, 28, 28, 1) / 255.0
X_train
```

```python
X_normal = X_train[y_train != 24]
```

```python
model = Sequential()
model.add(Conv2D(32, (3,3), activation='relu', padding='same', input_shape=(28,28,1)))
model.add(MaxPooling2D((2,2)))
model.add(Conv2D(16, (3,3), activation='relu', padding='same'))
model.add(MaxPooling2D((2,2)))

model.add(Conv2D(16, (3,3), activation='relu', padding='same'))
model.add(UpSampling2D((2,2)))
model.add(Conv2D(32, (3,3), activation='relu', padding='same'))
model.add(UpSampling2D((2,2)))
model.add(Conv2D(1, (3,3), activation='sigmoid', padding='same'))
```

```python
model.compile(optimizer='adam',loss='mean_squared_error')
```

```python
r=model.fit(X_normal,X_normal,epochs = 11,batch_size=128,validation_data=(X_test,X_test))
```

```python
predicted = model.predict(X_test)

mse = np.mean(np.square(X_test-predicted),axis=(1,2,3))
```

```python
normal_predicted = model.predict(X_normal)
normal_mse = np.mean(np.square(X_normal - normal_predicted), axis=(1,2,3))
threshold = np.percentile(normal_mse, 95)
```

```python
outliers = (mse > threshold).astype(int)
actual = (y_test==24).astype(int)
```

```python
print(confusion_matrix(actual,outliers))
print(classification_report(actual,outliers))
```
