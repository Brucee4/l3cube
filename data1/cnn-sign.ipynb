```python
# a. Import necessary packages

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout


# b. Load and preprocess image data

train = pd.read_csv("sign_mnist_train.csv")
test = pd.read_csv("sign_mnist_test.csv")

x_train = train.drop(['label'], axis=1).values
y_train = train['label'].values

x_test = test.drop(['label'], axis=1).values
y_test = test['label'].values

# Convert labels from 0-24 to 0-23
y_train = np.where(y_train > 9, y_train - 1, y_train)
y_test = np.where(y_test > 9, y_test - 1, y_test)

# Set image shape and number of classes
image_shape = (28, 28, 1)
num_classes = 24

# Normalize data
x_train = x_train.astype('float32') / 255
x_test = x_test.astype('float32') / 255

# Reshape images
x_train = x_train.reshape(-1, *image_shape)
x_test = x_test.reshape(-1, *image_shape)


# c. Define CNN model architecture

model = Sequential()

model.add(Conv2D(32, (3,3), activation='relu', input_shape=image_shape))
model.add(MaxPooling2D((2,2)))

model.add(Conv2D(64, (3,3), activation='relu'))
model.add(MaxPooling2D((2,2)))

model.add(Flatten())
model.add(Dense(64, activation='relu'))
model.add(Dropout(0.2))
model.add(Dense(num_classes, activation='softmax'))

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()


# d. Train the model

history = model.fit(
    x_train, y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2
)


# e. Estimate model performance

test_loss, test_accuracy = model.evaluate(x_test, y_test)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy * 100, "%")


# Plot accuracy

plt.plot(history.history['accuracy'], label='Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('CNN Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()


# Plot loss

plt.plot(history.history['loss'], label='Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('CNN Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
```
