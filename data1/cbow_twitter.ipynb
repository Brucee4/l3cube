```python
import pandas as pd
import numpy as np
import re

from gensim.models import Word2Vec
from sklearn.model_selection import train_test_split
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("/home/vansh-talyani/Downloads/LP IV Updated Datasets/CBOW/twitter_dataset.csv")
```

```python
print(df['Text'].head(10).to_string())
```

```python
documents = df['Text'].dropna().astype(str).tolist()
documents[0]
```

```python
sentences = [re.findall(r'\w+',text.lower()) for text in documents]
```

```python
sentences[0]
```

```python
x_train, x_test,train_text,test_text = train_test_split(sentences,documents,test_size=0.2,random_state=42)
```

```python
w2v = Word2Vec(x_train,vector_size=50,window=2,min_count=1,sg=0,epochs=20)
```

```python
print(w2v.wv.most_similar("congress"))
```

```python
def get_vectors(words):
    vectors = []
    for w in words:
        if w in w2v.wv:
            vectors.append(w2v.wv[w])

    if len(vectors)>0:
        return np.mean(vectors,axis=0)
    else:
        return np.zeros(50)

x_train = np.array([get_vectors(s) for s in x_train])
x_test = np.array([get_vectors(s) for s in x_test])
```

```python
user_input = "congress party"

input_words = re.findall(r'\w+', user_input.lower())
input_vector = get_vectors(input_words).reshape(1, -1)
```

```python
similarities = cosine_similarity(input_vector, x_train)[0]
print(similarities)
```

```python
top_indices = np.argsort(similarities)[-5:][::-1]
```

```python
for i in top_indices:
    print(train_text[i])
    print(similarities[i])
```
